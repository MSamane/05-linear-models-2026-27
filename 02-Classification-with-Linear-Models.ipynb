# 5.3 Linear Models for Classification: Introduction to Logistic Regression

So far we have introduced linear models tailored for regression problems (i.e., problems where the 'answer' is a numeric value, like porosity). This is a relatively intuitive affair. But how do we apply linear models to classification tasks? We can do this by using the simplest of all classification models: Logistic Regression.

But exactly what *is* logistic regression? Logistic regression is based on the logistic function, which is a sigmoid function. You can see its shape in the figure below: on the left, we apply a fit to categorical data (which can take values of either 0 or 1, also known as a binary classification problem) using a simple linear regression model, which we can clearly see is not a good fit. However, if we use the logistic function, we get a much better fit!




![logit](img/log_lin.png)<br>

The logistic function (aka 'sigmoid function') can be defined mathematically as:<br>
$$\sigma(x) = \frac{1}{1+e^{-x}}$$


If you look at the curve, you can see that what it does is map values such that: 

$$x_{-\infty}^{+\infty} \Rightarrow \sigma(x)_{0}^{1}$$<br>

In non-mathematical terms, this means that the sigmoid function (represented by $\sigma$) returns a value between y = 0 and y = 1 depending on the value of x. 

<a href='https://medium.com/@ODSC/logistic-regression-with-python-ede39f8573c7'>Source</a><br>

$\color{green}{ \sigma(x)}$ is therefore a probability ($p$) and can be rewritten as:<br>
$$p = \sigma(X) = {\frac{1}{1+e^{-(\beta_0 + \beta_1 X_1 + ... + \beta_n X_n)}}}$$
But how is this a **linear model**?

### 5.3.1 Odds, log odds,and logit space

The first step to making this a linear function is to use **odds** rather than **probability**. Specifically, we want to find the odds of the probability that a record belongs to class 1, rather than the probability of the record belonging to class 1. We can define the odds of the probability **($p$)** as:

$$\text{odds}=\frac{p}{1-p}$$

This may seem a little confusing, but it's much more intuitive if you 'say' it out loud:

 - If the probability of sample A belonging to class 1 is 50%, then the odds that it belongs to class 1 are 1:1.
 - If the probability of sample A belonging to class 1 is 67%, then the odds that it belongs to class 1 are 2:1.

Some of you might understand this better if we plot it out for a range of probabilities:

```python
import matplotlib.pyplot as plt
import numpy as np

# Odds curve
p = np.arange(0.01,1,0.005)
odds = p / (1-p)
plt.plot(p, odds); plt.ylabel("Odds"); plt.xlabel("p");
```

We can see that as the probability tends to 1, the odds tend to infinity. 

We can then define the natural logarithm of the odds (log odds, or "logit") in the form of a linear model:

$$\ln(\frac{p}{(1-p)}) = \text{logit}(p)=\beta_0 + \beta_1 X_1 + ... + \beta_n X_n$$

This means that we can use the logit function to fit a linear model in order to find the probability (p) that a record belongs to a given class. After we can find the parameters to help us define the logit function for our dataset, we can draw a function showing the relationship between the logit unit and the probability (an example of this curve is found below). Consequently, if we solve for the logit, we can calculate the probability of the sample belonging to a given class:

```python
# log-odds = LOGIT
plt.plot(p, np.log(odds)); plt.grid()
plt.title('Logit Function'); plt.minorticks_on()
plt.grid(which='minor', linestyle=':', linewidth='0.5', color='black')
plt.grid(which='major', linestyle='-', linewidth='0.5', color='black')
plt.xlabel('proba'); plt.ylabel('logit')
plt.scatter(0.9, np.log(0.9/(1-0.9)), c='red');
```


<p>👉 In other words, the logistic regression algorithm maps probability range <code>[0-1]</code> to log odds <code>[-inf, +inf]</code></p>

There are two possible types of "error" for classification problems: either the error is **False Positive** (Type I error; the model thinks the observation is Positive when its actually Negative), or it can be a **False Negative** (Type II error; the model thinks the observation is Negative when its actually Positive). The best way to represent classification errors is through the use of a tool called a confusion matrix. 

### Confusion Matrix
In a confusion matrix, the real class (sometimes presented as the True Labels) is generally represented on the vertical axis, while the Predicted class is represented on the horizontal axis. It thus becomes immediately apparent where the model has misclassified the data. A 'perfect' model would have all of its predictions falling on the main diagonal of the matrix (i.e., no False Positives or False Negatives), but this is pretty much impossible.

![conf_mat1](img/confusion_matrix_editable.png)

## Accuracy
The first metric that we can use for classification is the one that first comes to mind: accuracy. Accuracy is the sum of the correct predictions divided by the sum of the overall number of predictions.

$$\text{accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$

![cm](img/confusion_matrix_worked_example.png)

$\text{accuracy} = \frac{4 + 3}{4 + 3 + 2 + 1} = 0.7$<br>

<p>👉 <a href="https://scikit-learn.org/stable/modules/generated/sklearn.metrics.accuracy_score.html#sklearn.metrics.accuracy_score">Sklearn's <code>accuracy_score</code> documentation</a></p>


### Limitations of accuracy

Accuracy has a significant number of limitations, especially when the dataset is imbalanced (i.e., when not all classes are represented equally). Let's assume that you have created a 'model' that always predicts a 'Negative' value. Take a look at the following dataset, and at the accuracy that your 'model' has:

![cm](img/confusion_matrix_accuracy_warning.png)


$accuracy = \frac{TP + TN}{TP + TN + FP + FN} = \frac{0 + 99}{0 + 99 + 1 + 0} = 0.99$

🚨Using accuracy alone can give overly confident scores, especially when dealing with imbalanced datasets🚨

## Recall
Recall measures the ability of the model to detect occurrences of a class.

$$recall = \frac{TP}{TP + FN}$$

![cm](img/recall-annotated.png)

$\text{recall=}\frac{4}{4 + 1}=0.8$<br>

👉 <a href="https://scikit-learn.org/stable/modules/generated/sklearn.metrics.recall_score.html#sklearn.metrics.recall_score">Sklearn's <code>recall_score</code> documentation</a>


### Limitations of recall
Recall is not always the right measure for your model. For instance, see what happens in the case where you only detect a small portion of the true positives:

![cm](img/recall_warning.png)

In this example, the model detects 27 of the 30 actual positive cases:
$recall = \frac{TP}{TP + FN} = \frac{27}{27 + 3} = 0.90$

You should use recall when:
* Having a large number of false positives is an acceptable outcome for your classification task, e.g., cancer identification
* An example where recall would be a poor metric is when you're attempting to detect smoke for a smoke alarm - I'm sure many of you have been annoyed when your smoke alarm goes off when you're cooking food! This is a classic example of a False Positive.

However, it also incorrectly flags 100 negative cases as positive. Its precision is therefore only 27 / (27 + 100) ≈ 0.21. This shows why we should consider precision alongside recall.

## Precision
Precision measures the ability of a model to correctly identify TPs.

$$precision = \frac{TP}{TP + FP}$$

![cm](img/precision-annotated.png)

$\text{precision}=\frac{4}{4 + 2} = 0.67$<br>

👉 <a href="https://scikit-learn.org/stable/modules/generated/sklearn.metrics.precision_score.html#sklearn.metrics.precision_score">Sklearn's <code>precision_score</code> documentation</a>

### Limitations of precision
Precision has the inverse problem of recall: it focuses strongly on true positives at the expense of true negatives:

![cm](img/precision_problem.png)

$\text{precision} = \frac{TP}{TP + FP} = \frac{356}{356 + 44} = 0.89$

$\text{recall} = \frac{TP}{TP + FN} = \frac{356}{356 + 900} = 0.28$

## $F_1-score$
The F1-score is combines precision and recall into a single metric.

$$F_1=2x\frac{precision \times recall}{precision + recall}$$

* It is also known as the harmonic mean of precision and recall
* It will be influenced more by the lower of the two values

![cm](img/confusion_matrix_worked_example.png)

$\text{F}_1\text{ score}=2 ·\frac{0.67 \times 0.8}{0.67 + 0.8} = 0.73$<br>

👉 <a href="https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html#sklearn.metrics.f1_score">Sklearn's <code>f1_score</code> documentation</a>


In summary, here is what each classification metric gives you:
* 👉 Use **Accuracy** when you have balanced classes and predicting each class is important
* 👉 Use **Recall** when it is important to identify as many occurrences of a class as possible.
* 👉 Use **Precision** when it is important to correctly identify the positive class.
* 👉 Use **$F_1$** when you want a generic metric to compare across models and datasets

## Creating a Confusion Matrix

As you've seen above, confusion matrices are the clearest way of communicating the performance of your classification algorithm to both yourself and others. Thankfully, `sklearn` has a really simple function that allows to construct a confusion matrix in the form of a numpy array.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

data = pd.read_csv('./Data/core_data.csv')

data.drop_duplicates(inplace = True)

X = data.drop(columns = 'Porosity (vol%)')
y = data['Porosity (vol%)'].apply(lambda x: np.where(x >= 53, 1, 0))

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.7, random_state = 42)

pipe = Pipeline([
    ('imputer', SimpleImputer()),
    ('scaler', MinMaxScaler()),
    ('model', LogisticRegression())
])

pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score

# Generates a NxN confusion matrix, where N is the number of classes being predicted
# Output is a numpy array
cm = confusion_matrix(y_test, y_pred)

cm
```

We can improve the visualisation of this confusion matrix by using the Seaborn heatmap plot as follows. You can, of course, improve the visualisation by adding a title, adjusting the tick labels on the x- and y-axis to be more descriptive, etc. Feel free to experiment!

```python
import seaborn as sns

# Create a heatmap.fmt = 'g' prevents seaborn from using scientific notation on this plot
sns.heatmap(cm, annot = True, cmap = 'coolwarm', fmt = 'g')

plt.xlabel('Predicted Value', fontsize = 14, weight = 'bold')
plt.ylabel('Actual Value', fontsize = 14, weight = 'bold');
```
